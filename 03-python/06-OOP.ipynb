```python
numbers = [1, 2, 3, 4]

def solution(numbers):
   total = sum(numbers)
   average = total / len(numbers)
   answer = round(average, 1)

   return answer

result = solution(numbers)
print(result)
```

```python
def solution(year):
    if year % 400 == 0:
        answer = 'YES'
    elif year % 100 == 0:
        answer = 'NO'
    elif year % 4 == 0:
        answer = 'YES'
    else:
        answer = 'NO'
    
    return answer

print(solution(1900))
```

```python
def solution(n, numbers):
    answer = 0

    for number in numbers:
        if number == n:
            answer += 1

    return answer

result = solution (5, [5, 5, 5, 5, 5, 5, 5])
print(result)

```
